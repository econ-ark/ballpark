## Algan and Ragot (2009): Bellman Stage Decomposition

This notebook rewrites the household problem from Algan and Ragot (2009) in the “stage / perch” language used in class. The model is first laid out, and then its Bellman equation is decomposed into two within-period stages plus the between-period connector.

### 1. The paper and its model

**Paper.** Algan, Yann, and Xavier Ragot (2009): *Monetary policy with heterogeneous agents and borrowing constraints*.

**Objective.** A unit mass of ex-ante identical, infinitely lived households chooses consumption, real money balances, labor supply, and a risky asset position to maximize expected lifetime utility in an incomplete-markets production economy with borrowing constraints and monetary policy.

**Preferences.** For a household $i$, period utility is
$$
 u(c_t^{i}, m_t^{i}, \, l_t^{i}) = \frac{1}{1-\sigma} \left[ \left( \omega (c_t^{i})^{\frac{\eta-1}{\eta}} + (1-\omega) (m_t^{i})^{\frac{\eta-1}{\eta}} \right)^{\frac{\eta}{\eta-1}} (1-l_t^{i})^{\psi} \right]^{1-\sigma},
$$
with standard parameters $\omega, \eta, \psi, \sigma$.

**Idiosyncratic risk.** Each household faces an idiosyncratic labor productivity shock $e_t^{i} \in \{e_h,e_m,e_l\}$ following a 3-state Markov chain. Labor income is $w_t e_t^{i} l_t^{i}$.

**Assets and borrowing constraints.** Households can save in:

- **Real money** $m_t^{i}$
- **A risky asset** $a_{t+1}^{i}$ yielding return $r_t$

There is a borrowing constraint limiting $a_{t+1}^{i}$ (and effectively net wealth), so some agents are constrained and others are unconstrained.

**Government and transfers.** The government raises seigniorage revenue via inflation and redistributes it as transfers $\mu_t^{i}$. There are proportional taxes on capital and labor income. Monetary policy follows a rule for the real money stock $\Omega_t$; inflation tax revenue
$$
 \tau_t^{tot} = \pi_t \frac{\Omega_{t-1}}{\Pi_t}
$$
enters the government budget and finances $\mu_t^{i}$ and public spending.

**State variables (household level).** At the arrival perch for period $t$:

- **Financial/resources state:** $q_t^{i}$ a consolidated measure of beginning-of-period net resources (wealth) for household $i$, summarizing past asset choices and returns.
- **Idiosyncratic productivity state:** $e_t^{i}$.

We treat $q_t^{i}$ and $e_t^{i}$ as the **core endogenous state** at the start of the period. Aggregate states (capital, money, prices, tax rates) are taken as given by the individual.

**Control (choice) variables within a period.** Given $(q_t^{i}, e_t^{i})$, the household chooses:

- Consumption $c_t^{i}$
- Real money balances $m_t^{i}$
- Labor supply $l_t^{i}$
- Next-period risky asset position $a_{t+1}^{i}$

These are subject to a within-period budget constraint and the borrowing constraint.

**Within-period budget constraint (schematic).** Suppressing taxes for brevity and focusing on the household-level problem:
$$
 c_t^{i} + m_t^{i} + a_{t+1}^{i} = q_t^{i} + w_t e_t^{i} l_t^{i} + \mu_t^{i},
$$
with an implicit constraint $a_{t+1}^{i} \geq \underline{a}$ capturing borrowing limits.

**Bellman equation (paper’s monolithic formulation).** The household’s problem can be written as:
$$
 V(q_t^{i}, e_t^{i})
 = \max_{c_t^{i},\,m_t^{i},\,l_t^{i},\,a_{t+1}^{i}} 
 \Big\{ u(c_t^{i}, m_t^{i}, l_t^{i}) + \beta \, \mathbb{E}\big[ V(q_{t+1}^{i}, e_{t+1}^{i}) \mid e_t^{i} \big] \Big\}.
$$

The law of motion for $q_{t+1}^{i}$ is determined by asset returns, inflation, taxes, and transfers; we make it explicit when defining the connector.

Here $V(\cdot)$ is the **Bellman value function**, with **multiple control variables** $(c_t^{i}, m_t^{i}, l_t^{i}, a_{t+1}^{i})$ chosen simultaneously in the monolithic formulation.

### 2. Why multiple stages?

The monolithic Bellman equation treats all choices $(c_t, m_t, l_t, a_{t+1})$ as being made **simultaneously** given the state $(q_t, e_t)$. However, the economic structure of the model naturally suggests **sequential within-period decisions** that can be represented as distinct stages with their own arrival, decision, and continuation perches.

There are at least two conceptually separate decisions:

- **Labor-supply decision.** Given current productivity $e_t$ and resources $q_t$, the household decides how much labor $l_t$ to supply. This choice trades off leisure against labor income $w_t e_t l_t$ and therefore affects current-period utility through the leisure term $(1-l_t)^{\psi}$ and expands the budget set for the subsequent portfolio and consumption decisions.
- **Portfolio–consumption decision.** Conditional on the **resources available after labor income and transfers** (call this $x_t$), the household allocates those resources between immediate consumption $c_t$, real money balances $m_t$, and risky asset holdings $a_{t+1}$. This is a multi-control portfolio choice problem between a liquid asset (money), an illiquid/risky asset, and contemporaneous consumption.

These decisions are tightly **coupled** in the monolithic problem because labor supply both changes current utility and shifts resources available for portfolio and consumption. But economically it is natural to think of the household as first choosing labor supply (given wages and productivity) and then, **given the implied resources**, choosing how to allocate those resources across assets and consumption.

From a DDSL point of view, this suggests a decomposition into at least **two within-period stages**:

1. A **labor-supply stage** that takes $(q_t,e_t)$ as the arrival state, chooses $l_t$, and passes forward updated resources $x_t$.
2. A **consumption/portfolio stage** that takes $(x_t,e_t)$ as the arrival state and chooses $(c_t,m_t,a_{t+1})$.

Decomposing the problem this way has several advantages:

- **Clarifies timing:** It makes explicit that productivity $e_t$ is observed at the beginning of the period (arrival perch), labor is chosen in response, and then residual resources are allocated across assets and consumption.
- **Exposes structure of first-order conditions:** The labor-supply FOC involves the shadow value of resources entering the consumption/portfolio stage. The staged formulation isolates this shadow value as the marginal value of $x_t$ in the second stage.
- **Modularity:** The consumption/portfolio stage is itself a multi-control problem, but it can be studied and solved conditional on $x_t$ and $e_t$, potentially using specialized numerical methods. The labor-supply stage then wraps around it as an outer optimization.
- **Easier to extend:** Additional features (e.g. separate liquid and illiquid assets, adjustment costs, money-in-utility refinements) can often be inserted as additional stages without rewriting the entire Bellman equation.

### 3. Stage decomposition

We now decompose the monolithic Bellman problem into two within-period stages plus an inter-period connector:

- **Stage 1: Labor-supply stage**
- **Stage 2: Consumption/portfolio stage**
- **Connector:** Maps end-of-period asset holdings and money balances into next period’s state.

#### Stage summary

| Stage name              | Arrival state          | Decision variables          | Continuation state              | Notes                                      |
|-------------------------|------------------------|-----------------------------|----------------------------------|--------------------------------------------|
| `labor-supply`          | $(q_t, e_t)$           | $l_t$                       | $(x_t, e_t)$                    | Labor income $w_t e_t l_t$ added to $q_t$. |
| `consumption-portfolio` | $(x_t, e_t)$           | $c_t, m_t, a_{t+1}$         | $(a_{t+1}, m_t, e_t)$           | Subject to $c_t + m_t + a_{t+1} = x_t$.    |
| connector               | $(a_{t+1}, m_t, e_t)$  | —                           | $(q_{t+1}, e_{t+1})$            | Applies returns, inflation, transitions.   |

Throughout, we suppress aggregate states and prices in the notation, treating them as parameters from the household’s perspective.

#### 3.1. Stage 1: Labor-supply stage

**Name.** `labor-supply` stage.

**Arrival perch (labor-supply stage).**

At the beginning of period $t$, after inter-period returns have been realized but before any within-period decisions are taken, the household arrives at the labor-supply stage with state
$$
 S_t^{LS} = (q_t, e_t),
$$
where $q_t$ summarizes beginning-of-period resources and $e_t$ is the realized productivity shock.

All relevant shocks for period $t$ at the household level (here, $e_t$) are **already realized at this arrival perch**. There are no additional within-period shocks in the model’s baseline specification.

**Decision perch (labor-supply stage).**

Given $(q_t, e_t)$, the household chooses labor supply $l_t \in [0,1]$. This decision affects:

- Current-period utility via the leisure term $(1-l_t)^{\psi}$ inside the period utility function.
- Available resources for subsequent consumption/portfolio decisions via labor income $w_t e_t l_t$.

We can write the **labor-supply stage value** as:
$$
 V^{LS}(q_t,e_t)
 = \max_{l_t \in [0,1]} \Big\{ u^{LS}(l_t; q_t,e_t) + V^{CP}(x_t, e_t) \Big\},
$$
where:

- $u^{LS}(l_t; q_t,e_t)$ collects the part of period utility attributable to leisure, abstracting from $(c_t,m_t)$ for now.
- $V^{CP}(x_t, e_t)$ is the value of the **consumption/portfolio stage**, to be defined below, conditional on the resources $x_t$ that emerge from the labor-supply decision.

**Continuation perch (labor-supply stage).**

The continuation state that exits this stage and enters the consumption/portfolio stage is
$$
 S_t^{CP} = (x_t, e_t),
$$
where
$$
 x_t = q_t + w_t e_t l_t + \mu_t
$$
collects all resources available for allocation between consumption, money, and next-period asset holdings after labor income and transfers are added.

Thus, the **continuation mapping** from labor supply $l_t$ to the arrival state of the next stage is
$$
 (q_t,e_t) \xrightarrow{l_t} (x_t,e_t) \quad\text{with}\quad x_t = q_t + w_t e_t l_t + \mu_t.
$$

#### `labor-supply` stage perches

| Perch        | Description                                                                 |
|--------------|-----------------------------------------------------------------------------|
| Arrival      | Observe $(q_t, e_t)$ (all idiosyncratic shocks for period $t$ realized).    |
| Decision     | Choose labor supply $l_t \in [0,1]$.                                        |
| Continuation | Pass $(x_t, e_t)$, where $x_t = q_t + w_t e_t l_t + \mu_t$, to next stage.  |

#### 3.2. Stage 2: Consumption/portfolio stage

**Name.** `consumption-portfolio` stage.

**Arrival perch (consumption/portfolio stage).**

The consumption/portfolio stage receives as its arrival state
$$
 S_t^{CP} = (x_t, e_t),
$$
where $x_t$ is the resources available after labor income and transfers, constructed in the previous stage. No new shocks are realized at this stage in the baseline model; $e_t$ was already observed at the labor-supply stage arrival.

**Decision perch (consumption/portfolio stage).**

Given $(x_t,e_t)$, the household chooses:

- Consumption $c_t \ge 0$
- Real money balances $m_t \ge 0$
- Next-period risky asset holdings $a_{t+1} \ge \underline{a}$

subject to the **resource constraint**
$$
 c_t + m_t + a_{t+1} = x_t.
$$

The value of the consumption/portfolio stage is
$$
 V^{CP}(x_t,e_t)
 = \max_{c_t, m_t, a_{t+1}} \Big\{ u^{CP}(c_t,m_t; e_t) + \beta \, \mathbb{E}\big[ V(q_{t+1}, e_{t+1}) \mid e_t, a_{t+1}, m_t \big] \Big\}
$$
subject to
$$
 c_t + m_t + a_{t+1} = x_t, \quad a_{t+1} \ge \underline{a}.
$$

Here, the expectation over $V(q_{t+1}, e_{t+1})$ reflects **inter-period dynamics** mediated by the connector, which takes $(a_{t+1}, m_t, e_t)$ as inputs.

**Continuation perch (consumption/portfolio stage).**

At the end of this stage, before moving to the next period, the continuation state is
$$
 S_t^{\text{end}} = (a_{t+1}, m_t, e_t),
$$
which contains all information needed by the connector to generate next period’s state $(q_{t+1}, e_{t+1})$. No further within-period decisions are made after this stage.

#### `consumption-portfolio` stage perches

| Perch        | Description                                                                                         |
|--------------|-----------------------------------------------------------------------------------------------------|
| Arrival      | Receive $(x_t, e_t)$ from `labor-supply` stage.                                                     |
| Decision     | Choose $c_t \ge 0$, $m_t \ge 0$, $a_{t+1} \ge \underline{a}$ subject to $c_t + m_t + a_{t+1} = x_t$. |
| Continuation | Pass $(a_{t+1}, m_t, e_t)$ to the connector.                                                        |

#### 3.3. Inter-period connector

The **connector** is not itself a stage; it maps the end-of-period continuation state into the next period’s arrival state, integrating over any shocks realized at the start of period $t+1$.

**Inputs to the connector (end of period $t$).**

From the consumption/portfolio stage, we have:

- Risky asset position: $a_{t+1}$
- Real money balances: $m_t$
- Current productivity shock: $e_t$ (relevant for transition probabilities)

Aggregate states (capital stock, money stock, inflation, gross returns, tax rates, etc.) are also inputs, but we suppress them in the individual’s notation.

**Connector law of motion.**

The connector applies the asset return and inflation/tax structure to produce next period’s beginning-of-period resources $q_{t+1}$. Schematically,
$$
 q_{t+1} = f\big(a_{t+1}, m_t; \text{prices}_t, \text{policy}_t\big).
$$
At the same time, the idiosyncratic productivity shock evolves according to a Markov transition matrix.

Thus, from the perspective of the consumption/portfolio stage, the connector induces the conditional expectation
$$
 \mathbb{E}\big[ V(q_{t+1}, e_{t+1}) \mid a_{t+1}, m_t, e_t \big].
$$

**Outputs of the connector (arrival state for period $t+1$).**

The connector outputs next period’s arrival state for the labor-supply stage:
$$
 S_{t+1}^{LS} = (q_{t+1}, e_{t+1}).
$$

No choices are made and no additional shocks are realized “between periods”; all uncertainty is handled either at the arrival perches of stages or inside the connector.

#### 3.4. Composition of stages and connector

Putting stages and connector together, a full period can be written as the composition:
$$
 V(q_t,e_t)
 = V^{LS}(q_t,e_t)
 = \max_{l_t} \Big\{ u^{LS}(l_t; q_t,e_t) + V^{CP}(x_t,e_t) \Big\},
$$
where
$$
 x_t = q_t + w_t e_t l_t + \mu_t
$$
comes from the continuation mapping of the labor-supply stage, and
$$
 V^{CP}(x_t,e_t)
 = \max_{c_t,m_t,a_{t+1}} \Big\{ u^{CP}(c_t,m_t; e_t)
 + \beta \, \mathbb{E}\big[ V(q_{t+1}, e_{t+1}) \mid a_{t+1}, m_t, e_t \big] \Big\}
$$
subject to
$$
 c_t + m_t + a_{t+1} = x_t, \quad a_{t+1} \ge \underline{a}.
$$

In DDSL language, the period-$t$ Bellman operator is composed **right-to-left** as
$$
 V = \texttt{labor-supply} \circ \texttt{consumption-portfolio} \circ \texttt{connector}.
$$

### 4. Comparison with the monolithic formulation

The original monolithic Bellman equation
$$
 V(q_t, e_t)
 = \max_{c_t,m_t,l_t,a_{t+1}} \Big\{ u(c_t, m_t, l_t) + \beta \, \mathbb{E}[ V(q_{t+1}, e_{t+1}) ] \Big\}
$$
subject to
$$
 c_t + m_t + a_{t+1} = q_t + w_t e_t l_t + \mu_t
$$
can be recovered by:

1. Substituting $x_t = q_t + w_t e_t l_t + \mu_t$ into the resource constraint,
2. Recognizing that the outer maximization over $l_t$ in the labor-supply stage and the inner maximization over $(c_t,m_t,a_{t+1})$ in the consumption/portfolio stage together span the same feasible set of $(c_t,m_t,l_t,a_{t+1})$ as the monolithic problem,
3. Recombining $u^{LS}$ and $u^{CP}$ into the full utility function $u(c_t,m_t,l_t)$.

The staged formulation clarifies how the **labor-supply channel** and the **portfolio channel** separately contribute to the overall optimization, and it isolates the **shadow value of resources** that links the two stages. It also makes transparent where the borrowing constraint $a_{t+1} \ge \underline{a}$ bites (in the consumption/portfolio stage) and how it interacts with inflation and transfers through the connector.
