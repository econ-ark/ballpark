## Monetary Growth Literature

The paper builds on the monetary growth literature by relaxing the representative-agent assumption. Key foundational works include:

### Tobin Effect

{cite:t}`tobin1965money` identified the **portfolio substitution effect** (Tobin effect), where individuals switch from holding money to capital when inflation increases. This is the conventional channel that operates when there is no borrowing constraint.

**Key mechanism:** Inflation reduces the real return on money, making capital relatively more attractive, leading to increased capital accumulation.

### Phelps Effect

{cite:t}`phelps1973inflation` showed the **negative impact of inflation tax on other distorting taxes**, specifically capital and labor taxes. When inflation tax revenues are used to reduce distortionary taxes, this can have positive effects on capital accumulation.

**Key mechanism:** Inflation tax can substitute for more distortionary taxes, reducing deadweight loss and potentially increasing capital accumulation.

## Incomplete Markets Literature

### Aiyagari (1994)

{cite:t}`aiyagari1994uninsured` provides the foundational framework for heterogeneous-agent models with incomplete markets. The model shows how:

- Idiosyncratic income risk leads to precautionary saving
- Borrowing constraints affect portfolio choices
- Aggregate capital accumulation differs from representative-agent models

**Connection to Algan and Ragot (2009):** The paper embeds money into an Aiyagari-style incomplete-markets production economy, allowing monetary policy to interact with precautionary saving motives.

### Chari, Christiano, and Kehoe (2000)

{cite:t}`chari2000sticky` developed the CES utility function specification that Algan and Ragot adopt for money-in-the-utility-function models.

**Utility specification:** The CES form allows for flexible interest elasticity of money demand, which is crucial for analyzing portfolio substitution effects.

## Heterogeneous-Agent Monetary Models

Prior to Algan and Ragot (2009), most monetary models assumed representative agents. The paper's key innovation is introducing:

1. **Heterogeneity** - Different households face different borrowing constraints
2. **Binding constraints** - Some households are constrained, others are not
3. **Asymmetric responses** - Constrained and unconstrained households respond differently to inflation

## Theoretical Foundations

The paper combines insights from:

- **Monetary growth theory** (Tobin, Phelps) - Portfolio substitution and fiscal effects
- **Incomplete markets theory** (Aiyagari) - Precautionary saving and borrowing constraints  
- **Heterogeneous-agent macroeconomics** - Distributional effects and aggregate implications

This synthesis allows the paper to identify novel transmission mechanisms that are absent in representative-agent models.
