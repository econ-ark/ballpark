## Guner, Kaygusuz, and Ventura (2012) — Ballpark Entry

**Paper:** Nezih Guner, Remzi Kaygusuz, and Gustavo Ventura, "Taxing Women: A Macroeconomic Analysis," *Journal of Monetary Economics* 59(1), 111–128, 2012. [DOI: 10.1016/j.jmoneco.2011.10.002](https://doi.org/10.1016/j.jmoneco.2011.10.002)

**Why in-ballpark:** The paper is a serious structural evaluation of a specific policy conjecture (lower marginal tax rates on married women) inside a calibrated life-cycle general-equilibrium model with married and single households, permanent productivity types, endogenous human capital, and a log-progressive tax function. Its headline quantitative finding — that while gender-based taxation does improve welfare over the status-quo joint-filing system, a gender-**neutral** proportional income tax dominates both — is a sharp, counter-intuitive result that exemplifies what a well-identified HA-OLG tax-reform exercise can say. The model admits a clean modular-DDSL decomposition whose consumption-savings stage is EGM-amenable under log utility, making it a tractable REMARK candidate.
