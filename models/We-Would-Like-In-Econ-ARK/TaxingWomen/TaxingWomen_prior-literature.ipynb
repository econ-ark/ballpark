# Prior Literature

## Overview

Guner, Kaygusuz, and Ventura (2012, hereafter GKV 2012) sits at the intersection of two literatures. The first is **optimal income taxation**, beginning with {cite:t}`Mirrlees1971-pd` and its family-tax extensions: {cite:t}`Boskin1983-wd` showed that equal marginal taxation of spouses is generally not optimal when secondary earners are more labor-supply elastic, and {cite:t}`Kleven2009-xx` derived the general nonlinear optimal income tax for couples, proving that the marginal tax on secondary earnings should decline with primary earnings. A closely related contemporaneous contribution, {cite:t}`Alesina2011-km`, gives the direct normative case for taxing women at lower rates than men on Ramsey grounds when women's labor supply is more elastic. These papers motivate the welfare question the paper asks.

The second pillar is the **quantitative life-cycle literature on female labor supply**. {cite:t}`Attanasio2008-cn` provides the life-cycle framework with an operative extensive margin and heterogeneity in child-related costs and gender wage gaps; this is the empirical backbone for quantitative evaluation of gender-differentiated tax reforms. The authors' own prior contribution, {cite:t}`Guner2012-ty`, embeds these mechanisms in a heterogeneous-household, life-cycle general-equilibrium model and evaluates proportional and separate-filing reforms — setting the stage for the present paper's focus on gender-differentiated rates.

## Key Foundational Papers

{cite:t}`Mirrlees1971-pd` — canonical optimal-income-tax framework; benchmark redistribution-efficiency trade-off that underlies all subsequent family-taxation analysis.

{cite:t}`Boskin1983-wd` — equal marginal taxation of spouses is not optimal; with higher secondary-earner elasticities, efficient schedules imply lower marginal rates for secondary earners.

{cite:t}`Kleven2009-xx` — general nonlinear optimal income tax for couples as a multidimensional screening problem; direct mechanism-design complement to GKV 2012’s quantitative evaluation.

{cite:t}`Alesina2011-km` — contemporaneous normative companion: gender-based taxation (lower marginal rates on women) satisfies Ramsey's rule when women's labor supply is more elastic, with microfoundations from intra-household bargaining.

{cite:t}`Attanasio2008-cn` — life-cycle quantitative evidence on female labor supply; identifies the roles of child-related costs and the gender wage gap in shaping participation dynamics across cohorts.

{cite:t}`Guner2012-ty` — the authors' own *REStud* precursor: heterogeneous-household, life-cycle general-equilibrium model with an operative extensive margin; proportional-filing and separate-filing reforms.

## The Gap This Paper Addresses

Optimal tax theory ({cite:t}`Mirrlees1971-pd`; {cite:t}`Boskin1983-wd`; {cite:t}`Kleven2009-xx`) establishes *that* secondary earners should face lower marginal rates than primary earners, and {cite:t}`Alesina2011-km` maps this to gender when labor-supply elasticities are gendered. {cite:t}`Attanasio2008-cn` provides the life-cycle empirical framework with child-cost heterogeneity. {cite:t}`Guner2012-ty` supplies the calibrated heterogeneous-household general-equilibrium laboratory.

Missing is a quantitative evaluation of *gender-differentiated* rates — as distinct from separate filing or proportional reforms — inside a calibrated life-cycle general-equilibrium model with married and single households, children, human-capital accumulation, and a log-progressive tax function. GKV 2012 fills this gap and produces the counter-intuitive quantitative result that, while gender-based taxation does improve welfare over the status-quo joint-filing system, a gender-*neutral* proportional income tax dominates both.
