# Subsequent Literature

## Research Directions Since Guner, Kaygusuz, and Ventura (2012)

Since the publication of {cite:t}`Guner2012-xi`, the literature on taxation and household labor supply has expanded in several directions. A major theme has been cross-country measurement and modeling of couple-level tax wedges and secondary-earner responses. {cite:t}`taxation_bick_2018` provides a comparative macro framework across countries, while {cite:t}`longterm_bick_2018` documents long-run changes in married couples' labor supply and taxes since the 1980s.

A second strand has focused on embedding family taxation inside richer policy environments. Rather than analyzing filing status in isolation, newer work studies bundled reforms that include social insurance, child-related transfers, and labor-supply incentives. Examples include {cite:t}`social_kaygusuz_2015`, {cite:t}`joint_nishiyama_2019`, and {cite:t}`childrelated_guner_2020`.

The most visible post-2020 quantitative follow-up in the paper's design space is {cite:t}`marriage_borella_2023`, which asks directly whether U.S. marriage-related taxes and Social Security rules depress female labor supply. Using a life-cycle model of married and single households with human-capital accumulation, they find that moving from joint filing to separate filing and from spousal and survivor Social Security benefits to individual benefits raises participation and hours of married women by about 20\% at prime ages — a quantitative magnitude comparable to the gender-neutral proportional reform in {cite:t}`Guner2012-xi` and consistent with its core insight that secondary-earner margins are first-order for tax design.

A third direction links gender gaps, time allocation, and macro outcomes. The synthesis in {cite:t}`families_doepke_2016` (*Handbook of Macroeconomics*, chapter 23) provides the canonical survey of how family structure, intra-household allocation, bargaining, and fertility interact with macroeconomic aggregates, and contextualizes {cite:t}`Guner2012-xi` within the broader "families in macroeconomics" research program. {cite:t}`allocation_duernecker_2018` emphasizes household time-use and production margins; {cite:t}`evolution_olivetti_2016` synthesizes cross-country evidence on gender inequality.

## Key Subsequent Papers

{cite:t}`taxation_bick_2018` — cross-country quantitative benchmark for how tax-and-transfer systems shape married couples' labor supply, especially along the secondary-earner margin.

{cite:t}`marriage_borella_2023` — most influential post-2020 follow-up in the paper's design space: life-cycle structural evaluation of how U.S. marriage-related taxes and Social Security benefits together depress female labor supply, with large counterfactual gains from joint-to-separate filing and individual Social Security benefits.

{cite:t}`families_doepke_2016` — *Handbook of Macroeconomics* chapter providing the canonical survey of family structure, intra-household allocation, and fertility in macroeconomic models; situates {cite:t}`Guner2012-xi` within the broader "families in macroeconomics" agenda.

{cite:t}`childrelated_guner_2020` — child-related transfers, taxes, and labor-supply responses analyzed jointly; shifts the policy frontier from single-instrument reforms to integrated family-policy design.

{cite:t}`social_kaygusuz_2015` — how Social Security rules interact with two-earner household decisions; life-cycle channels for labor supply and welfare.

{cite:t}`joint_nishiyama_2019` — joint labor-supply responses to pension-system incentives; importance of modeling couples' coordinated decisions.

## Open Questions and Gaps

Despite substantial progress, several questions remain open.

**Intra-household decision structure.** Much of the quantitative literature still relies on simplified household decision rules, even though bargaining power, home production, and childcare frictions are central for identifying gender-specific elasticities and optimal wedges.

**Measurement and identification.** Translating statutory tax schedules into effective couple-level marginal tax rates remains difficult, and reform-based identification of extensive-margin responses is often limited by data comparability across countries and over time.

**Unified policy design.** A key open issue is how to jointly design income taxation, child-related transfers, Social Security, and childcare policy when participation, fertility, and household time allocation interact dynamically — the direction taken by {cite:t}`marriage_borella_2023` and {cite:t}`childrelated_guner_2020` but not yet fully integrated.

Overall, subsequent research confirms the core insight in {cite:t}`Guner2012-xi`: secondary-earner behavior is macro-relevant. The frontier now is to integrate richer household mechanisms with cleaner empirical identification so policy counterfactuals are both structurally coherent and data-disciplined.
