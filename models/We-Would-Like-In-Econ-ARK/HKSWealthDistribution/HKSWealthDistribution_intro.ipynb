**Pitch — why this paper is in the ballpark**

- *What the paper uniquely does.* Hubmer, Krusell, and Smith provide the first quantitative heterogeneous-agent theory that jointly accounts for the level of U.S. top-wealth concentration and its 1967–2012 dynamics, decomposing the rise into contributions from declining tax progressivity, rising earnings risk, and wealth-dependent portfolio returns.
- *Why Econ-ARK cares.* The model's Gaussian-AR(1) stochastic discount factor and wealth-indexed return schedules extend the standard Bewley–Huggett–Aiyagari machinery already in HARK with mechanisms that have become central to modern inequality work, and the paper's channel decomposition (Table 2) disciplines which of them matter quantitatively.
- *What a REMARK would enable.* Counterfactual experiments on tax progressivity, return-heterogeneity shutdowns, and extensions along the post-HKS research lines surfaced in [`HKSWealthDistribution_subsequent-literature.ipynb`](HKSWealthDistribution_subsequent-literature.ipynb) (monetary-policy transmission, micro-data discipline on saving and returns, market-power and belief channels).
