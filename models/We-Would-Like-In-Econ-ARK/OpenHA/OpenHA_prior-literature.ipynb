This notebook surveys the key prior literature on which {cite:t}`Auclert2021-ki` builds. The paper rests on three strands: open-economy New Keynesian foundations, heterogeneous-agent monetary transmission, and the exchange rate–real income–redistribution nexus.

## 1. Open-Economy New Keynesian Foundation

### Galí and Monacelli (2004)

{cite:t}`Gali2004-kj` lay out a small-open-economy version of the Calvo sticky-price model and reduce the equilibrium dynamics to a simple representation in domestic inflation and the output gap. {cite:t}`Auclert2021-ki` adopts this framework as its open-economy backbone and extends it along a single dimension — replacing the representative household with a continuum of incomplete-markets households — making Galí–Monacelli the single most structurally proximate antecedent.

## 2. Heterogeneous-Agent Monetary Transmission and the Intertemporal-MPC Apparatus

### Kaplan, Moll, and Violante (2018)

{cite:t}`Kaplan2018-to` revisit the transmission mechanism from monetary policy to household consumption in a Heterogeneous Agent New Keynesian (HANK) model. They show that **indirect** effects of interest rate cuts — operating through general equilibrium increases in labor demand — far outweigh **direct** intertemporal substitution effects. {cite:t}`Auclert2021-ki` adopts this direct-vs-indirect decomposition and extends it to an open-economy setting, where an additional channel — the real-income channel operating through import prices — enters as a third horizontal term.

### Auclert, Rognlie, and Straub (2024) — *The Intertemporal Keynesian Cross*

{cite:t}`Auclert2024-wc` establishes the **intertemporal MPC (iMPC) matrix $\mathbf{M}$** as the central sufficient-statistic object for the demand block of heterogeneous-agent New Keynesian models: each column $\mathbf{M}_{\cdot,s}$ is the impulse response of aggregate consumption to a dated income shock $s$ periods in the future, and $\mathbf{C} = \mathbf{M}\,\mathbf{Y}$ closes the Keynesian cross in sequence space. This is the **same object** that appears in the OpenHA decomposition $d\mathbf{C} = -\tfrac{\alpha}{1-\alpha}\mathbf{M}\,d\mathbf{Q} + \mathbf{M}\,d\mathbf{Y}$ in {cite:t}`Auclert2021-ki`; reading this paper is the clearest way to understand why OpenHA maps cleanly onto HARK's sequence-space Jacobian interface.

### Auclert (2019) — Monetary Policy and the Redistribution Channel

{cite:t}`Auclert2019-js` identifies three channels through which monetary policy redistributes — an earnings heterogeneity channel, a Fisher channel from unexpected inflation, and an interest-rate-exposure channel — and shows in Italian and US data that all three amplify monetary transmission. OpenHA extends this redistribution logic to the open-economy context: exchange-rate movements create a **fourth** redistributive effect (through import prices and the real-income channel) that is absent in a closed economy.

## 3. Exchange Rates, Real Income, and Heterogeneous-Agent Open-Economy Bridges

### Krugman and Taylor (1978)

{cite:t}`Krugman1978-yz` develop a simple Keynesian model showing that currency depreciation can be **contractionary** if imports exceed exports, if there are differences in consumption propensities out of profits and wages, or if government revenues rise with devaluation. This early formalization of the contractionary-depreciation hypothesis is precisely the phenomenon OpenHA is built to **size**: {cite:t}`Auclert2021-ki` identifies the MPC-differential mechanism Krugman and Taylor sketch with the real-income channel and asks how large it actually is in a modern HA calibration.

### de Ferra, Mitman, and Romei (2020)

{cite:t}`de-Ferra2020-ai` show how household heterogeneity — in particular, heterogeneous portfolios and exposure to currency mismatches — alters the transmission of foreign shocks in an open economy with incomplete markets. Their work is the most direct HA-open-economy antecedent to {cite:t}`Auclert2021-ki`; OpenHA differs principally in isolating the *real-income* channel and quantifying it against the expenditure-switching channel, rather than focusing on portfolio exposure.

## Further reading

Papers that are influential in neighboring literatures but not on the structural critical path of OpenHA. Listed without `{cite:t}` markers to keep the main survey to the 3–6 anchors above (per CONTRIBUTING.md's Primer criterion); full entries live in `references.bib`.

**Open-economy / NOEM foundations (broader New Keynesian antecedents of the Galí–Monacelli closure).**

- Obstfeld and Rogoff (1995), "Exchange Rate Dynamics Redux" — two-country GE with nominal rigidities; launched New Open Economy Macroeconomics. (`Obstfeld1995-wg`)
- Clarida, Galí, and Gertler (2002), "A Simple Framework for International Monetary Policy Analysis" — two-country sticky-price analysis of cooperation vs. Nash. (`Clarida2002-gd`)
- Corsetti and Pesenti (2001), "Welfare and Macroeconomic Interdependence" — choice-theoretic model of structural spillovers. (`Corsetti2001-dc`)

**Incomplete-markets building blocks (theoretical antecedents of the HA block).**

- McKay, Nakamura, and Steinsson (2016), "The Power of Forward Guidance Revisited" — precautionary-savings attenuation of forward-guidance power under incomplete markets. (`McKay2016-ss`)
- Werning (2015), "Incomplete Markets and Aggregate Demand" — generalized Euler relation in incomplete-markets economies. (`Werning2015-vz`)

**Exchange-rate redistribution and contractionary-depreciation lineage.**

- Díaz-Alejandro (1963), "A Note on the Impact of Devaluation and the Redistributive Effect" — early statement of the MPC-differential redistribution mechanism; subsumed into Krugman–Taylor's formalization for our purposes. (`Diaz-Alejandro1963-zf`)
- Cravino and Levchenko (2016), "The Distributional Consequences of Large Devaluations" — empirical heterogeneity of exchange-rate effects across consumption baskets. (`Cravino2016-hl`)
- Bems and di Giovanni (2016), "Income-Induced Expenditure Switching" — micro-evidence from Latvian scanner data that income effects, not relative prices, drive much of the import compression. (`Bems2016-us`)

**Computational/numerical bridge to HARK.**

- Auclert, Bardóczy, Rognlie, and Straub (2021), "Using the Sequence-Space Jacobian to Solve and Estimate Heterogeneous-Agent Models" — the computational apparatus that makes the iMPC matrix $\mathbf{M}$ numerically tractable and is the direct bridge between OpenHA and HARK's sequence-space Jacobian interface. (`Auclert2021-jz`)

## Key Gap Addressed

Prior work had either open-economy models with a real income channel but no household heterogeneity, or HANK models with redistribution but no open-economy structure. The real income channel had therefore never been **sized** in a model with both ingredients. {cite:t}`Auclert2021-ki` builds a tractable open-economy HANK and uses it to quantify the real income channel — including its redistribution and MPC dimensions — relative to the standard expenditure switching channel.
