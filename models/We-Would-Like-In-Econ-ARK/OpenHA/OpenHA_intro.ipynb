## Why this paper is in-ballpark for Econ-ARK

*What the paper uniquely does.* {cite:t}`Auclert2021-ki` embed an incomplete-markets, heterogeneous-agent household block inside the canonical small-open-economy New Keynesian framework and **size** the real-income channel of exchange-rate shocks — showing that at sufficiently low trade elasticity it can dominate the expenditure-switching channel and deliver contractionary depreciations.

*Why Econ-ARK cares.* The central decomposition $d\mathbf{C} = -\tfrac{\alpha}{1-\alpha}\,\mathbf{M}\,d\mathbf{Q} + \mathbf{M}\,d\mathbf{Y}$ rests on the matrix $\mathbf{M}$ of intertemporal MPCs — exactly the sequence-space Jacobian object HARK's incomplete-markets household block already exposes — so OpenHA is the cleanest open-economy mapping for HARK's existing HA consumption block.

*What a REMARK promotion would enable.* A reproducible decomposition of output and consumption impulse responses into real-income, multiplier, and expenditure-switching channels, calibrated side by side for an advanced-economy ($\alpha$, $\chi$ à la US) and an emerging-market ($\alpha$, $\chi$ à la Brazil) parameterization.
