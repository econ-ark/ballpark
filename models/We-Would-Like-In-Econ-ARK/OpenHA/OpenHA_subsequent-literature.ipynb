# Subsequent Literature Citing OpenHA

This notebook surveys papers that cite Auclert, Rognlie, Souchier, and Straub (2021). The citing work clusters into three broad directions: open-economy HANK in emerging markets, extensions beyond the small open economy, and fiscal–monetary interaction.

## 1. Open-Economy HANK in Emerging Markets

Several papers apply and extend the real income channel of Auclert, Rognlie, Souchier, and Straub (2021) to emerging-market settings, where contractionary devaluations, dollarization, and high MPC heterogeneity are first-order concerns.

### Campos and Paz (2023)

Campos and Paz (2023) study the real income channel and contractionary devaluations in a heterogeneous-agent model calibrated to Latin American economies. They show that the mechanisms identified by Auclert, Rognlie, Souchier, and Straub (2021) — the real income and multiplier channels — are quantitatively large in countries with high import dependence and concentrated income distributions.

### Blanco, Drenik, and Zaratiegui (2025)

Blanco, Drenik, and Zaratiegui (2025) study how nominal devaluations affect inflation and inequality. Building on the open-economy HANK framework of Auclert, Rognlie, Souchier, and Straub (2021), they show that devaluations disproportionately hurt lower-income households through import price increases, and that this distributional channel feeds back into aggregate dynamics.

### Ferrante and Gornemann (2022)

Ferrante and Gornemann (2022) examine how deposit dollarization interacts with household heterogeneity during devaluations. They extend the framework of Auclert, Rognlie, Souchier, and Straub (2021) by adding a dollarized financial sector, showing that balance-sheet effects from currency mismatch amplify the contractionary real income channel.

### Signe, Poutineau, and Gankou (2026)

Signe, Poutineau, and Gankou (2026) apply heterogeneous-agent monetary policy analysis to the CEMAC zone (Central Africa), drawing on Auclert, Rognlie, Souchier, and Straub (2021) to study how exchange rate policy interacts with household heterogeneity in developing-country monetary unions.

### Oskolkov (2021)

Oskolkov (2021) studies exchange rate policy and heterogeneity in small open economies, examining how the optimal exchange rate regime depends on the degree of household heterogeneity — a question directly motivated by the findings of Auclert, Rognlie, Souchier, and Straub (2021).

## 2. Beyond the Small Open Economy

A second strand extends Auclert, Rognlie, Souchier, and Straub (2021) from a small open economy to two-region or monetary-union HANK models, and studies international spillovers.

### Bayer, Kriwoluzky, Müller, and Seyrich (2024)

Bayer, Kriwoluzky, Müller, and Seyrich (2024) develop a HANK² model of monetary unions — a two-region heterogeneous-agent model where each region has incomplete markets. They build on Auclert, Rognlie, Souchier, and Straub (2021) to show how asymmetric shocks and heterogeneous fiscal spaces interact within a currency union.

### Bellifemine, Couturier, and Jamilov (2025)

Bellifemine, Couturier, and Jamilov (2025) study monetary unions with heterogeneous fiscal space. Extending the open-economy HANK approach of Auclert, Rognlie, Souchier, and Straub (2021) to a multi-country setting, they show that differences in fiscal capacity across member states alter the transmission of common monetary policy.

### Yang, Zhang, and Hou (2023)

Yang, Zhang, and Hou (2023) develop a two-country HANK model with trade frictions, extending the single-country framework of Auclert, Rognlie, Souchier, and Straub (2021) to study how household heterogeneity in both countries affects the international transmission of shocks.

### Chen, Lazarakis, and Varthalitis (2025)

Chen, Lazarakis, and Varthalitis (2025) study debt targets and fiscal consolidation in a Euro Area HANK model, incorporating the open-economy heterogeneous-agent mechanisms from Auclert, Rognlie, Souchier, and Straub (2021) to analyze how fiscal rules interact with monetary policy in a currency union.

### Acharya and Pesenti (2024)

Acharya and Pesenti (2024) study international spillovers and spillbacks of monetary policy. Drawing on Auclert, Rognlie, Souchier, and Straub (2021), they show how heterogeneous-agent channels amplify the cross-border transmission of monetary policy shocks.

### Georgiadis, Müller, and Schumann (2024)

Georgiadis, Müller, and Schumann (2024) study global risk and the dollar, examining how US monetary policy shocks spill over to other economies through exchange rate and real income channels identified by Auclert, Rognlie, Souchier, and Straub (2021).

### Kyriazis (2023)

Kyriazis (2023) studies quantitative easing spillovers, analyzing how unconventional monetary policy transmits internationally through the heterogeneous-agent channels emphasized by Auclert, Rognlie, Souchier, and Straub (2021).

### De Leo, Gopinath, and Kalemli-Özcan (2022)

De Leo, Gopinath, and Kalemli-Özcan (2022) study monetary policy cyclicality in emerging economies, showing that the real income channel from Auclert, Rognlie, Souchier, and Straub (2021) helps explain why emerging-market central banks often tighten rather than loosen in response to capital outflows.

## 3. Fiscal–Monetary Interaction

A third cluster of papers uses the framework of Auclert, Rognlie, Souchier, and Straub (2021) to study interactions between fiscal and monetary policy in open economies.

### Aggarwal, Auclert, Rognlie, and Straub (2023)

Aggarwal, Auclert, Rognlie, and Straub (2023) study excess savings and twin deficits — the transmission of fiscal stimulus in open economies. Building directly on Auclert, Rognlie, Souchier, and Straub (2021), they show that fiscal transfers in a HANK open economy generate both current account deficits (through higher imports) and budget deficits, with the magnitudes depending on household heterogeneity and the real income channel.

### Auclert, Monnery, Rognlie, and Straub (2023)

Auclert, Monnery, Rognlie, and Straub (2023) study how to manage an energy shock with fiscal and monetary policy in an open-economy HANK model. They extend Auclert, Rognlie, Souchier, and Straub (2021) to analyze supply-side shocks, showing that the real income channel is central to understanding how energy price increases transmit to aggregate demand.

### Nuño, Renner, and Scheidegger (2025)

Nuño, Renner, and Scheidegger (2025) study monetary policy with persistent supply shocks, incorporating the heterogeneous-agent transmission mechanisms from Auclert, Rognlie, Souchier, and Straub (2021) to analyze how central banks should respond when supply disruptions have distributional consequences.

### McKay and Wolf (2022)

McKay and Wolf (2022) study optimal policy rules in HANK models. They extend the analysis to open-economy settings influenced by Auclert, Rognlie, Souchier, and Straub (2021), examining how the presence of the real income channel and MPC heterogeneity alters the design of optimal monetary policy rules.

### Pieroni (2023)

Pieroni (2023) studies energy shortages and aggregate demand in a HANK model, showing that the output loss and distributional burden from energy shocks are amplified by the real income channel mechanisms identified in Auclert, Rognlie, Souchier, and Straub (2021).

## Open Questions

Cutting-edge topics building on Auclert, Rognlie, Souchier, and Straub (2021) include:

- Empirical validation of the real income channel with micro data
- Integrating household and firm heterogeneity in one open-economy HANK
- A unified treatment of financial frictions and dollarization with the real income channel
- Non-linearities and large shocks (contractionary devaluations, hysteresis)
- Optimal exchange rate regime choice in open-economy HANK
- Welfare-based policy evaluation that explicitly accounts for inequality and heterogeneity
