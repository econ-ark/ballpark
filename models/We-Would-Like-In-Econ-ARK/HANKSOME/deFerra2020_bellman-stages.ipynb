### Household heterogeneity and the transmission of foreign shocks

#### **1. The paper and its model**

**Paper:** @de-Ferra2019-co — [Household heterogeneity and the transmission of foreign shocks](https://doi.org/10.1016/j.jinteco.2020.103303), *Journal of International Economics*, 124, 103303, 2020.

**Authors:** Sergio de Ferra, Kurt Mitman, Federica Romei

##### **The agent’s objective**

The authors develop a heterogeneous-agent New Keynesian small open economy model (HANKSOME) to study how household heterogeneity affects the transmission of foreign shocks, particularly during a current account reversal. 
The model highlights how households' portfolio composition and the extent of foreign currency borrowing shape the contraction in consumption following sudden stops in capital inflows.

Households maximize the expected present discounted value of utility:
$$
U = E_0 \sum_{t=0}^{\infty} \beta^t \left[ u(c_t) - g(l_t) \right]
$$
where $\beta \in (0,1)$ is the household's subjective discount factor, $g(l_t)$ represents the disutility of supplying labor $l_t$, and $c_t$ is a CES aggregate of home and foreign goods.

Households' labor productivity $\{s_t\}_{t=0}^{\infty}$ is stochastic and is characterized by an $N$-state Markov chain that can take values $s_t \in S = \{s_1, \dots, s_N\}$ with transition probabilities given by $\gamma(s_{t+1} \mid s_t)$ and $\int s = 1$. Households rent their labor services, $l_t s_t$, to firms for a real wage $w_t$ per effective hour. $W_t$ denotes the corresponding nominal wage.

Agents rent their nominal asset portfolio $a_t$ to the asset market and earn a nominal return $i_t^a$. Households can hold multiple asset classes, including capital, real bonds, and nominal bonds denominated in domestic and foreign currencies. The nominal assets carried into the next period are given by
$$
a_{t+1} = b_{H,t+1} + e_t b_{F,t+1} + P_{H,t} \left( q_t k_{t+1} + \tilde{b}_{H,t+1} \right),
$$
where $b_{H,t+1}$ and $b_{F,t+1}$ denote nominal bonds denominated in domestic and foreign currency, respectively, $e_t$ is the nominal exchange rate, $q_t$ is the real price of capital in units of the home good, and $\tilde{b}_{H,t+1}$ are real bonds denominated in units of the home good. 

An unexpected change in the exchange rate $\Delta e_t$ therefore induces a one-time change in household wealth through the revaluation of foreign-currency debt, which affects household consumption.

Households take prices and wages as given. 
At time $t$, an agent faces the following nominal budget constraint:
$$
P_t c_t + a_{t+1} = (1 + i_t^a)a_t + W_t l_t s_t + d_t
$$
where $P_t$ is the price of a unit of consumption in terms of domestic currency, or the consumer price index (CPI), and $d_t$ are dividends distributed by firms.

In addition, households are subject to a borrowing constraint:
$$
a_{t+1} \geq \bar{a}
$$
which implies that assets carried into the next period cannot fall below the threshold $\bar{a}$.

##### **State variables and control variables**

In the household problem, the **state variables** at the beginning of a period are asset holdings $a$ and labor productivity $s$, while the **control variable** within the period is consumption $c$. 
In this model, labor supply is not a household choice variable in the Bellman problem. Instead, households supply labor services to firms at a wage that is set optimally by the union. As a result, $l_t$ is not treated as a control variable in the household's dynamic optimization problem.

##### **Portfolio composition and exchange rate revaluation**

Although the paper models four asset classes—domestic nominal bonds $b_H$, foreign-currency bonds $b_F$, physical capital $k$, and real bonds $\tilde{b}_H$—individual households do not choose portfolio shares across these classes as a control variable. The composition of the representative portfolio, and hence the return $i^a_t$, is determined in general equilibrium. From the perspective of the individual household's Bellman problem, total assets $a$ and its return $i^a$ are scalar quantities, and no portfolio allocation stage appears in the household's dynamic optimization.

The exchange rate $e_t$ affects household wealth through the revaluation of foreign-currency bonds $e_t b_{F,t+1}$. An unexpected change $\Delta e_t$ acts as a one-time wealth shock. Since this revaluation is embedded in the return $i^a_t$ on total assets, it does not require a separate stage in the household's problem—it is absorbed into the returns computation within the shocks-only stage (see Section 3).

##### **Shocks**

An idiosyncratic labor productivity shock $s$ is realized at the beginning of each period and becomes part of the household's state before decisions are made. The shock follows a Markov process, so households form expectations over the next-period productivity state $s'$ when solving the Bellman equation.


##### **Bellman equation**

The household's Bellman equation is given by:
$$
V(a,s;\Omega)
=
\max_{c}
\left\{
u(c) - g(l)
+
\beta
\sum_{s' \in S}
\gamma(s' \mid s)
V(a', s'; \Omega')
\right\}
$$

subject to
$$
Pc + a' = (1 + i^a)a + Wls + \pi
$$

$$
\Omega' = \Upsilon(\Omega)
$$

$$
c \ge 0
$$

$$
a' \ge \bar{a}
$$
where $\Omega(a,s)$ is the distribution on the space $X = \mathcal{A} \times \mathcal{S}$ of agents' asset holdings $a \in \mathcal{A}$ and labor productivity $s \in \mathcal{S}$ across the population, which, together with policy variables, determines equilibrium prices. $\Upsilon$ is an equilibrium object that specifies the evolution of the distribution $\Omega$.







#### **2. Why multiple stages?**

At the beginning of each period, the household observes the current state, including asset holdings and the realized productivity shock. Given this realization, total resources are determined and the household chooses consumption $c$, subject to the budget constraint. Because of this constraint, the choice of $c$ determines next-period assets $a'$.

**This problem has one choice variable together with a non-trivial shock structure that generates multiple stages.**

Moreover, because labor supply $l_t$ is set by the union and not chosen by the household, the disutility $g(l_t)$ is a known additive term (conditional on aggregate state) in the household's period objective. It does not affect the first-order condition for consumption. The value function must still include $-g(l_t)$, but the consumption policy function is determined entirely by $u(c)$ and the continuation value.

Structuring the problem into stages clarifies the timing of shocks and decisions. In the DDSL framework, each stage becomes a module with well-defined perches, which makes the structure of the dynamic problem easier to understand, and facilitates its numerical solution and simulation.


#### **3. Stage decomposition**

Let 
$$
m_t = P_t c_t + a_{t+1} = (1 + i_t^a)a_t + W_t l_t s_t + \pi_t
$$
denote nominal market resources.

Once the productivity shock $s_t$ is realized, market resources $m_t$ are determined. 
Therefore, in the following analysis I use $m_t$ instead of $s_t$ as the relevant state variable, 
while leaving the other state variable unchanged.

Define 
$$
\psi_t = m_t - P_t c_t
$$
as assets after all actions are completed. (We do not use $k$ because $k$ denotes physical capital in this paper.) 

Next period's asset holdings are therefore given by
$$
a_{t+1} = \psi_t
$$

Both $\psi$ and $a$ are $\psi$-type variables, representing investable assets before returns are realized. In contrast, the $m$-type variables like $m$ represents spendable resources after returns.

##### **Normalization**

In [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/), all variables are normalized by permanent income $P\Gamma$, yielding dimensionless quantities that make the EGM and the inverse-marginal-value transformation straightforward. In @de-Ferra2019-co, there is no permanent income growth (productivity is a stationary Markov chain), so $\Gamma = 1$. However, the budget constraint is written in nominal terms with the CPI $P_t$ multiplying consumption: $P_t c_t + a' = m_t$. This introduces a $P_t$ factor into the first-order condition that is absent in the standard [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) formulation.

To restore the standard EGM form, one can normalize by $P_t$ within each period. Define $\tilde{m}_t = m_t / P_t$ and $\tilde{\psi}_t = \psi_t / P_t$, so that the budget constraint becomes $c_t + \tilde{\psi}_t = \tilde{m}_t$ and the first-order condition takes the standard form $u'(c) = \tilde{v}'_\succ(\tilde{\psi})$. In the stage tables below, we retain nominal variables for consistency with the paper, but note that normalization by $P_t$ is needed before applying the standard EGM algorithm.

##### **3.2 The Shocks-only Stage**
<div align="center">

**Table 2. Shocks-only Stage Perches**

</div>

| Perch        | Indicator | State | Value functions              | Explanation      |
|--------------|-----------|-------|------------------------------|------------------|
| Arrival      | $\prec$   | $a$   | $v_{\prec} = \mathbb{E}_{\prec}[v_{\succ}]$ | pre-shock value  |
| Decision(s)  | $\sim$    | $a$   | (none)                       | no choice        |
| Continuation | $\succ$   | $\check{m}$ | $v_{\succ}$                 | post-shock value |

n.b.: $\check{m}$ is $m$-type (market resources after productivity shocks), matching $m$. 

The Arrival value function takes the expectation over the productivity shocks: $v_{\prec}(a) = \mathbb{E}_{\prec}[v_{\succ}((1 + i_t^a)a + W l s + \pi)]$. 
The Continuation state $\check{m} = (1 + i_t^a)a + W l s + \pi$ is fully determined (non-stochastic) once the shocks are realized.

**Mapping to the [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) `portable` stage.** This shocks-only stage is a specialization of the `portable` stage in [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) with no portfolio optimization (`bar_Shr = 0`). The correspondence is:
- The return factor $(1 + i_t^a)$ plays the role of $R_{\text{port}} / \Gamma = R$ (with $\Gamma = 1$ and no risky-share choice).
- The labor income $W_t l_t s_t + \pi_t$ plays the role of the transitory shock $\theta$, where the randomness comes from the Markov state $s_t$.
- The equilibrium objects $i_t^a$, $W_t$, $l_t$, $\pi_t$ are taken as given by the household—only $s_t$ is stochastic from the household's perspective.
- The exchange rate revaluation is absorbed into $(1 + i_t^a)$, which reflects the portfolio return inclusive of currency gains/losses on foreign-currency bonds.

##### **3.3 The Shock-free consumption Stage (cons-noshocks)**
With shocks in the preceding stage, the consumption stage has Arrival state $m$ and no shocks between Arrival and Decision, so $v_{\prec} = v_{\sim}$.

<div align="center">

**Table 3. Cons-noshocks Stage Perches**

</div>

| Perch        | Indicator | State | value functions | Explanation |
|--------------|-----------|-------|-----------------|-------------|
| Arrival      | $\prec$   | $m$   | $v_{\prec} = v_{\sim}$ | no shocks; identity |
| Decision(s)  | $\sim$    | $m$   | $v_{\sim}(m) = \max_c u(c) - g(l) + v_{\succ}(m- Pc)$ | choose consumption |
| Continuation | $\succ$   | $\psi$   | $v_{\succ}$ | value at exit |

Since $g(l_t)$ does not depend on $c$, the first-order condition is $u'(c) = P_t \cdot v'_\succ(\psi)$ where $\psi = m - P_t c$. The $P_t$ factor distinguishes this from the standard [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) form $u'(c) = v'_\succ(m - c)$. After normalizing by $P_t$ (as discussed above), the standard EGM applies:
- The consumed function is $c_\succ(\tilde{a}) = (\tilde{v}'_\succ(\tilde{a}))^{-1/\rho}$
- The endogenous grid point is $\tilde{m} = \tilde{a} + c_\succ(\tilde{a})$

where $\tilde{a} = \psi / P_t$ and $\tilde{m} = m / P_t$.


##### **3.4 The three-stage period**

The shocksonly–consnoshocks period is defined by the stage list  [shocks-only, $C(\check{m} \leftrightarrow m)$, cons-noshocks, disc]:

| Element | Transition | Action |
|--------|-----------|-------|
| shocks-only | $a \rightarrow \check{m}$ | shocks realize (no choice) |
| $\mathcal{C}(\check{m} \leftrightarrow m)$ | $\check{m} \rightarrow m$ | rename |
| cons-noshocks | $m \rightarrow \psi$ | choose $c$ |
| disc |  | apply $\beta$ |

##### **3.5 Connectors for Period**

The within-period connectors are specified above. For the between-period connector, the period ends after the disc stage (a passthrough), so the exit state is $\psi$. The shocks-only stage of the next period arrives with state $a$. Therefore, the between-period connector is
$\mathcal{C}(\psi \leftrightarrow a)$, which is the identity $a_{t+1} = \psi_t$.

Note that this between-period connector is where the exchange rate revaluation takes effect in the paper's narrative: the nominal value of assets carried into period $t+1$ depends on the exchange rate $e_{t+1}$, which may differ from $e_t$. In the stage framework, this revaluation is captured by the return $i^a_{t+1}$ at the start of the next period's shocks-only stage, not by the connector itself.

The pipeline is:
$$
a 
\xrightarrow{\text{shocks-only}} 
\check{m} 
\xrightarrow{\text{cons-noshocks}} 
\psi 
\xrightarrow{\text{disc}} 
\text{exit}
$$




#### **4. Comparison with the monolithic formulation**
##### **4.1 How does your staged decomposition relate to the Bellman equation you wrote in Section 1?**
The staged decomposition is mathematically equivalent to the Bellman equation presented in Section 1, but it separates the optimization into sequential stages. In the Bellman formulation,
$$
V(a_t,m_t;\Omega) = \max_{c_t} \left\{ u(c_t) - g(l_t) + \beta\, E\left[ V(a_{t+1}, m_{t+1}; \Omega_{t+1}) \right] \right\}
$$

The current utility, continuation value, and expectation over future shocks appear in a single expression. In the staged decomposition, these components are implemented in separate modules: the shocks-only stage forms expectations over the productivity shock, the cons-noshocks stage determines $c_t$ and $a_{t+1}$, and the discounting stage applies the factor $\beta$ to the continuation value. Composed together, these stages reproduce the same value function as the monolithic Bellman equation.

##### **4.2 What does the decomposition make clearer that the monolithic formulation obscures?**
The staged decomposition clarifies that the realization of productivity determines market resources before the consumption decision is made, and that the budget constraint links consumption to next-period assets. In the monolithic Bellman equation, these timing relationships are less transparent.

##### **4.3 Are there alternative decompositions? Why did you choose this one?**
Yes. One can also consider a **consnoshocks–shocksonly** period, which is defined by the stage list  
[cons-noshocks, $C(\psi \leftrightarrow a)$, shocks-only, disc]:

| Element | Transition | Action |
|--------|-----------|-------|
| cons-noshocks | $m \rightarrow \psi$ | choose $c$ |
| $\mathcal{C}(\psi \leftrightarrow a)$ | $\psi \leftrightarrow a$ | rename |
| shocks-only | $a \rightarrow \check{m}$ | shocks realize (no choice) |
| disc |  | apply $\beta$ |

Under this decomposition, the between-period connector is $\mathcal{C}(\check{m} \leftrightarrow m)$.

I choose the **shocksonly–consnoshocks** period because:
1. It is more intuitive and consistent with the description of the model in the paper: households make consumption decisions after receiving wage income.
2. It has a concrete advantage for the EGM. In the shocksonly–consnoshocks ordering, the `cons-noshocks` stage receives a continuation value $v_\succ(\psi)$ that already has shocks integrated out by the preceding `shocks-only` stage. This means the first-order condition $u'(c) = P \cdot v'_\succ(\psi)$ can be inverted directly via the EGM without any inner expectation loop. In the reverse ordering (consnoshocks–shocksonly), the consumption FOC would involve a continuation value that still contains unresolved shocks, requiring numerical root-finding inside an expectation integral—precisely the inefficiency the EGM is designed to avoid.

##### **4.4 Normalization and the EGM**

The @de-Ferra2019-co budget constraint $P_t c_t + a' = m_t$ differs from the [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) canonical form $c + a = m$ by the price factor $P_t$. To apply the EGM and the inverse-marginal-value transformation (Section 6 of [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/)), one should normalize all flow and stock variables by $P_t$ within each period, yielding $\tilde{m} = m/P$, $\tilde{\psi} = \psi/P$. After normalization:

- The consumed function is $c_\succ(\tilde{a}) = (\tilde{v}'_\succ(\tilde{a}))^{-1/\rho}$, which is nearly linear and easy to interpolate (the key insight of the "Transformation" in [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/), Section 6).
- The endogenous grid is $\tilde{m} = \tilde{a} + c_\succ(\tilde{a})$.
- The borrowing constraint $a' \ge \bar{a}$ becomes $\tilde{\psi} \ge \bar{a}/P_t$, which varies with $P_t$ across periods in a transition path.

Since $\Gamma = 1$, the $\Gamma$-normalization used in [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) is trivial here, simplifying the implementation.

##### **4.5 Path toward implementation**

To turn this decomposition into working code using the [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) modular architecture:

1. **Define the Markov process for $s$.** Discretize using the Rouwenhorst or Tauchen method. This gives shock nodes $\{s_1, \ldots, s_N\}$ and transition matrix $\Gamma(s'|s)$.

2. **Implement the shocks-only stage as a specialization of `solve_portable`.** Set `bar_Shr = 0` (no portfolio choice). The shock distribution combines the Markov transition for $s$ with deterministic equilibrium objects $(i^a, W, l, \pi)$. The mapping from the `portable` stage's $k$-grid to post-shock $\check{m}$ is: $\check{m}(a, s) = (1 + i^a) a + W l s + \pi$.

3. **Use `solve_cons_noshocks` for the consumption stage.** After normalizing by $P_t$, this is the standard EGM step. The `a_grid` is built from the natural borrowing constraint $\bar{a}/P_t$ upward using the multi-exponential grid (`setup_grids_expMult`).

4. **Use `solve_disc` with $\beta$ and $\Gamma = 1$.**

5. **Compose stages using `solve_period`** with a custom `PeriodSpec` that lists `[portable, cons-noshocks, disc]`, matching the `PERIOD_CONS_ONLY` type in [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/).

6. **Backward iteration.** For a stationary equilibrium, iterate backward from a terminal condition (or use `solve_modular_infinite` with a convergence criterion on the consumption function). For transition dynamics, solve a finite-horizon sequence of periods with time-varying $(i^a_t, W_t, l_t, P_t, \pi_t)$.

7. **Condition on $s$.** Because productivity $s$ is a discrete Markov state (not i.i.d.), the backward pass must maintain separate value/policy functions for each $s$-state. This is the main structural difference from the [SolvingMicroDSOPs](https://llorracc.github.io/SolvingMicroDSOPs/) baseline, where transitory shocks are i.i.d. and integrated out within `portable`.
